```python
import os
import random
import shutil

DATASET_DIR = "/content/drive/MyDrive/ENC-PSL/Cours_Traitement_Image_Automatique_Option_ENC_2026/Litt-Jeunesse/project-5-at-2026-03-31-14-44-b4d4b327"
DATASET_OUTPUT_DIR = "dataset"
IMAGES_DIR = os.path.join(DATASET_DIR, "images")
LABELS_DIR = os.path.join(DATASET_DIR, "labels")

TRAIN_RATIO = 0.8

# Funzione per dividere il dataset in train e val, mantenendo la corrispondenza tra immagini e label
def split_dataset():
    images = [f for f in os.listdir(IMAGES_DIR) if f.endswith(".jpg") or f.endswith(".png")]
    random.shuffle(images)

    split_idx = int(len(images) * TRAIN_RATIO)
    train_files = images[:split_idx]
    val_files = images[split_idx:]

    for split, files in [("train", train_files), ("val", val_files)]:
        os.makedirs(f"{DATASET_OUTPUT_DIR}/images/{split}", exist_ok=True)
        os.makedirs(f"{DATASET_OUTPUT_DIR}/labels/{split}", exist_ok=True)

        for f in files:
            shutil.copy(os.path.join(IMAGES_DIR, f),
                        os.path.join(DATASET_OUTPUT_DIR, "images", split, f))

            label_file = f.replace(".jpg", ".txt").replace(".png", ".txt")
            shutil.copy(os.path.join(LABELS_DIR, label_file),
                        os.path.join(DATASET_OUTPUT_DIR, "labels", split, label_file))

split_dataset()
```
