```python
import pandas as pd
import matplotlib.pyplot as plt

# Carica il file CSV
df = pd.read_csv('5_char_for_novel.csv')

# Mappa dei valori
mappa_type = {
    0: 'adult',
    1: 'enfant',
    2: 'animal',
    3: 'poupée',
    4: 'objet',
    5: 'autre'
}

# Conta le occorrenze
conteggio_type = df['type'].value_counts().sort_index()
conteggio_type = conteggio_type.reindex(range(6), fill_value=0)

# Sostituisce gli indici numerici con le etichette
conteggio_type.index = conteggio_type.index.map(mappa_type)

# Grafico
plt.figure()
conteggio_type.plot(kind='bar')
plt.xlabel('Type')
plt.ylabel('Nombre de personnages')
plt.title('Répartition des valeurs de type')

plt.show()
```
